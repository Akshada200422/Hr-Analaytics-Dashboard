2. Find missing values


```python
df = pd.read_excel(r"/content/HR Data.xlsx")

print(df.head())
print(df.isnull().sum())
print(df.duplicated().sum())
```

3. Handle missing values


```python
df.dropna()
```

```python
df.fillna(0)
```

```python
df["Age"].fillna(df["Age"].mean(), inplace=True)
```

```python
df["Monthly Income"].fillna(df["Monthly Income"].median(), inplace=True)
```

4. Remove duplicates

```python
df.duplicated()
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

5. Correct data types

```python
df.dtypes

df["Age"] = df["Age"].astype(int)
df["Monthly Income"] = df["Monthly Income"].astype(float)
```

6. CRemove extra spaces

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

7. Convert text to consistent case

```python
df["Gender"] = df["Gender"].str.strip().str.title()
df["Department"] = df["Department"].str.strip().str.title()
df["Job Role"] = df["Job Role"].str.strip().str.title()
```

```python
print(df["Gender"].unique())
print(df["Department"].unique())
```

9. Replace inconsistent values

```python
print(df["Attrition"].value_counts())
print(df["Over Time"].value_counts())
print(df["Business Travel"].value_counts())
```

10. Check categorical columns

```python
for col in text_columns:
    print("\n", col)
    print(df[col].unique())
```

11. Check numerical columns for invalid values

```python
print(df["Age"].min())
print(df["Age"].max())

print(df["Monthly Income"].min())
print(df["Monthly Income"].max())

print(df["Distance From Home"].min())
print(df["Distance From Home"].max())
```
